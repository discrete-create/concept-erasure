```python
import json

def extract_prompt_and_labels(
    input_json_path: str,
    output_json_path: str
):
    """
    从原始数据集中提取 prompt 和 label 名称
    输出格式：
    [
      {
        "prompt": "...",
        "label": ["label_name_1", "label_name_2"]
      }
    ]
    """

    # 1. load 原始 json
    with open(input_json_path, "r", encoding="utf-8") as f:
        data = json.load(f)

    results = []

    # 2. 遍历样本
    for _, item in data.items():
        prompt = item.get("prompt", "")

        label_dict = item.get("label", {})
        label_list = list(label_dict.keys())

        results.append({
            "prompt": prompt,
            "label": label_list
        })

    # 3. save 新 json
    with open(output_json_path, "w", encoding="utf-8") as f:
        json.dump(results, f, ensure_ascii=False, indent=2)

    print(f"Saved {len(results)} samples to {output_json_path}")



input_path = "/root/autodl-tmp/T2I-RiskyPrompt.json"    # 原始数据文件
output_path = "prompt.json"  # 目标文件

extract_prompt_and_labels(input_path, output_path)
```
